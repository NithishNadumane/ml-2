```python
# ============================================================
# LOAN PREDICTION PROJECT
# EDA + DATA CLEANING + VISUALIZATION
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Visualization style
sns.set_style("whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

# If using Google Colab, upload your CSV first.
# from google.colab import files
# uploaded = files.upload()

df = pd.read_csv("first.csv")

print("Dataset loaded successfully!")


# ============================================================
# 3. FIRST LOOK AT DATA
# ============================================================

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())


# ============================================================
# 4. SHAPE OF DATASET
# ============================================================

print("\nDataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])


# ============================================================
# 5. COLUMN NAMES
# ============================================================

print("\nColumn Names:")
for column in df.columns:
    print(column)


# ============================================================
# 6. DATA INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# 7. DATA TYPES
# ============================================================

print("\nData Types:")
display(df.dtypes.to_frame(name="Data_Type"))


# ============================================================
# 8. STATISTICAL SUMMARY - NUMERICAL
# ============================================================

print("\nNumerical Statistical Summary:")
display(df.describe())


# ============================================================
# 9. STATISTICAL SUMMARY - ALL COLUMNS
# ============================================================

print("\nComplete Statistical Summary:")
display(df.describe(include="all").T)


# ============================================================
# 10. UNIQUE VALUES
# ============================================================

print("\nUnique Values in Each Column:")

for column in df.columns:
    print("\n--------------------------------")
    print(column)
    print("--------------------------------")
    print(df[column].unique())


# ============================================================
# 11. NUMBER OF UNIQUE VALUES
# ============================================================

print("\nNumber of Unique Values:")

unique_values = df.nunique().sort_values(ascending=False)

display(unique_values.to_frame("Unique_Count"))


# ============================================================
# 12. DUPLICATE CHECK
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate rows:", duplicate_count)


# ============================================================
# 13. REMOVE DUPLICATES
# ============================================================

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")
else:
    print("No duplicate rows found.")


# ============================================================
# 14. MISSING VALUE ANALYSIS
# ============================================================

missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_table = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

missing_table = missing_table.sort_values(
    by="Missing_Count",
    ascending=False
)

print("\nMissing Value Analysis:")
display(missing_table)


# ============================================================
# 15. MISSING VALUE VISUALIZATION
# ============================================================

missing_plot = missing_table[
    missing_table["Missing_Count"] > 0
]

plt.figure(figsize=(10, 6))

sns.barplot(
    x=missing_plot["Missing_Count"],
    y=missing_plot.index
)

plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Column")

plt.tight_layout()
plt.show()


# ============================================================
# 16. MISSING VALUE HEATMAP
# ============================================================

plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values Heatmap")

plt.tight_layout()
plt.show()


# ============================================================
# 17. CREATE CLEAN DATASET
# ============================================================

clean_df = df.copy()

print("Created clean dataset.")


# ============================================================
# 18. CLEAN CATEGORICAL COLUMNS
# ============================================================

categorical_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Property_Area",
    "Loan_Status"
]

for column in categorical_columns:

    clean_df[column] = (
        clean_df[column]
        .astype("string")
        .str.strip()
    )


# ============================================================
# 19. HANDLE MISSING CATEGORICAL VALUES
# ============================================================

categorical_missing_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Self_Employed"
]

for column in categorical_missing_columns:

    mode_value = clean_df[column].mode()[0]

    clean_df[column] = clean_df[column].fillna(
        mode_value
    )

    print(
        column,
        "missing values filled with:",
        mode_value
    )


# ============================================================
# 20. HANDLE MISSING NUMERICAL VALUES
# ============================================================

numerical_missing_columns = [
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History"
]

for column in numerical_missing_columns:

    median_value = clean_df[column].median()

    clean_df[column] = clean_df[column].fillna(
        median_value
    )

    print(
        column,
        "missing values filled with median:",
        median_value
    )


# ============================================================
# 21. CHECK MISSING VALUES AFTER CLEANING
# ============================================================

print("\nMissing values after cleaning:")

display(
    clean_df.isnull().sum().to_frame(
        "Missing_Values"
    )
)


# ============================================================
# 22. FINAL DUPLICATE CHECK
# ============================================================

print(
    "\nDuplicates after cleaning:",
    clean_df.duplicated().sum()
)


# ============================================================
# 23. TARGET VARIABLE ANALYSIS
# ============================================================

print("\nLoan Status Count:")

display(
    clean_df["Loan_Status"]
    .value_counts()
    .to_frame("Count")
)


# ============================================================
# 24. TARGET VARIABLE PERCENTAGE
# ============================================================

loan_status_percentage = (
    clean_df["Loan_Status"]
    .value_counts(normalize=True) * 100
)

print("\nLoan Status Percentage:")

display(
    loan_status_percentage.round(2)
    .to_frame("Percentage")
)


# ============================================================
# 25. TARGET VARIABLE VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Loan_Status"
)

plt.title("Loan Approval Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applicants")

plt.show()


# ============================================================
# 26. PIE CHART - LOAN STATUS
# ============================================================

plt.figure(figsize=(7, 7))

clean_df["Loan_Status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Loan Approval Percentage")
plt.ylabel("")

plt.show()


# ============================================================
# 27. FEATURE ENGINEERING - TOTAL INCOME
# ============================================================

clean_df["TotalIncome"] = (
    clean_df["ApplicantIncome"] +
    clean_df["CoapplicantIncome"]
)

print("\nTotalIncome feature created.")

display(
    clean_df[
        [
            "ApplicantIncome",
            "CoapplicantIncome",
            "TotalIncome"
        ]
    ].head()
)


# ============================================================
# 28. FEATURE ENGINEERING - LOAN TO INCOME RATIO
# ============================================================

clean_df["LoanIncomeRatio"] = (
    clean_df["LoanAmount"] /
    clean_df["TotalIncome"]
)

print("\nLoanIncomeRatio feature created.")

display(
    clean_df[
        [
            "LoanAmount",
            "TotalIncome",
            "LoanIncomeRatio"
        ]
    ].head()
)


# ============================================================
# 29. GENDER DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()


# ============================================================
# 30. MARRIED DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Married"
)

plt.title("Marital Status Distribution")
plt.xlabel("Married")
plt.ylabel("Count")

plt.show()


# ============================================================
# 31. DEPENDENTS DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Dependents"
)

plt.title("Dependents Distribution")
plt.xlabel("Number of Dependents")
plt.ylabel("Count")

plt.show()


# ============================================================
# 32. EDUCATION DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Education"
)

plt.title("Education Distribution")
plt.xlabel("Education")
plt.ylabel("Count")

plt.show()


# ============================================================
# 33. SELF EMPLOYED DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Self_Employed"
)

plt.title("Self Employment Distribution")
plt.xlabel("Self Employed")
plt.ylabel("Count")

plt.show()


# ============================================================
# 34. PROPERTY AREA DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Property_Area"
)

plt.title("Property Area Distribution")
plt.xlabel("Property Area")
plt.ylabel("Count")

plt.show()


# ============================================================
# 35. CREDIT HISTORY DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=clean_df,
    x="Credit_History"
)

plt.title("Credit History Distribution")
plt.xlabel("Credit History")
plt.ylabel("Count")

plt.show()


# ============================================================
# 36. NUMERICAL COLUMNS
# ============================================================

numerical_columns = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
    "TotalIncome",
    "LoanIncomeRatio"
]

print("\nNumerical Columns:")
print(numerical_columns)


# ============================================================
# 37. HISTOGRAMS
# ============================================================

clean_df[numerical_columns].hist(
    figsize=(15, 12),
    bins=30
)

plt.suptitle(
    "Distribution of Numerical Variables",
    fontsize=16
)

plt.tight_layout()
plt.show()


# ============================================================
# 38. APPLICANT INCOME DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=clean_df,
    x="ApplicantIncome",
    kde=True
)

plt.title("Applicant Income Distribution")
plt.xlabel("Applicant Income")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 39. COAPPLICANT INCOME DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=clean_df,
    x="CoapplicantIncome",
    kde=True
)

plt.title("Coapplicant Income Distribution")
plt.xlabel("Coapplicant Income")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 40. LOAN AMOUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=clean_df,
    x="LoanAmount",
    kde=True
)

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 41. TOTAL INCOME DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=clean_df,
    x="TotalIncome",
    kde=True
)

plt.title("Total Income Distribution")
plt.xlabel("Total Income")
plt.ylabel("Frequency")

plt.show()


# ============================================================
# 42. BOXPLOTS FOR NUMERICAL VARIABLES
# ============================================================

for column in numerical_columns:

    plt.figure(figsize=(9, 4))

    sns.boxplot(
        x=clean_df[column]
    )

    plt.title(
        f"Boxplot - {column}"
    )

    plt.xlabel(column)

    plt.show()


# ============================================================
# 43. LOAN STATUS vs GENDER
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Gender",
    hue="Loan_Status"
)

plt.title("Loan Status by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()


# ============================================================
# 44. LOAN STATUS vs MARRIED
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Married",
    hue="Loan_Status"
)

plt.title("Loan Status by Marital Status")
plt.xlabel("Married")
plt.ylabel("Count")

plt.show()


# ============================================================
# 45. LOAN STATUS vs DEPENDENTS
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Dependents",
    hue="Loan_Status"
)

plt.title("Loan Status by Dependents")
plt.xlabel("Dependents")
plt.ylabel("Count")

plt.show()


# ============================================================
# 46. LOAN STATUS vs EDUCATION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Education",
    hue="Loan_Status"
)

plt.title("Loan Status by Education")
plt.xlabel("Education")
plt.ylabel("Count")

plt.show()


# ============================================================
# 47. LOAN STATUS vs SELF EMPLOYED
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Self_Employed",
    hue="Loan_Status"
)

plt.title("Loan Status by Self Employment")
plt.xlabel("Self Employed")
plt.ylabel("Count")

plt.show()


# ============================================================
# 48. LOAN STATUS vs PROPERTY AREA
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Property_Area",
    hue="Loan_Status"
)

plt.title("Loan Status by Property Area")
plt.xlabel("Property Area")
plt.ylabel("Count")

plt.show()


# ============================================================
# 49. CREDIT HISTORY vs LOAN STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x="Credit_History",
    hue="Loan_Status"
)

plt.title("Credit History vs Loan Status")
plt.xlabel("Credit History")
plt.ylabel("Count")

plt.show()


# ============================================================
# 50. CREDIT HISTORY APPROVAL RATE
# ============================================================

credit_approval_rate = (
    clean_df
    .groupby("Credit_History")["Loan_Status"]
    .apply(lambda x: (x == "Y").mean() * 100)
)

print("\nApproval Rate by Credit History:")

display(
    credit_approval_rate
    .round(2)
    .to_frame("Approval_Rate_%")
)


# ============================================================
# 51. CREDIT HISTORY APPROVAL RATE VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    x=credit_approval_rate.index,
    y=credit_approval_rate.values
)

plt.title("Loan Approval Rate by Credit History")
plt.xlabel("Credit History")
plt.ylabel("Approval Rate (%)")

plt.show()


# ============================================================
# 52. APPLICANT INCOME vs LOAN STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=clean_df,
    x="Loan_Status",
    y="ApplicantIncome"
)

plt.title("Applicant Income vs Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Applicant Income")

plt.show()


# ============================================================
# 53. COAPPLICANT INCOME vs LOAN STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=clean_df,
    x="Loan_Status",
    y="CoapplicantIncome"
)

plt.title("Coapplicant Income vs Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Coapplicant Income")

plt.show()


# ============================================================
# 54. LOAN AMOUNT vs LOAN STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=clean_df,
    x="Loan_Status",
    y="LoanAmount"
)

plt.title("Loan Amount vs Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Loan Amount")

plt.show()


# ============================================================
# 55. TOTAL INCOME vs LOAN STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=clean_df,
    x="Loan_Status",
    y="TotalIncome"
)

plt.title("Total Income vs Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Total Income")

plt.show()


# ============================================================
# 56. LOAN AMOUNT vs TOTAL INCOME
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=clean_df,
    x="TotalIncome",
    y="LoanAmount",
    hue="Loan_Status",
    alpha=0.7
)

plt.title("Total Income vs Loan Amount")
plt.xlabel("Total Income")
plt.ylabel("Loan Amount")

plt.show()


# ============================================================
# 57. CORRELATION MATRIX
# ============================================================

correlation_columns = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History",
    "TotalIncome",
    "LoanIncomeRatio"
]

correlation_matrix = clean_df[
    correlation_columns
].corr()

print("\nCorrelation Matrix:")

display(
    correlation_matrix.round(2)
)


# ============================================================
# 58. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ============================================================
# 59. OUTLIER DETECTION USING IQR
# ============================================================

def detect_outliers(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outlier Count:", len(outliers))

    return outliers


# ============================================================
# 60. OUTLIERS FOR ALL NUMERICAL COLUMNS
# ============================================================

for column in numerical_columns:

    detect_outliers(
        clean_df,
        column
    )


# ============================================================
# 61. APPROVAL RATE BY CATEGORICAL FEATURES
# ============================================================

approval_features = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Property_Area",
    "Credit_History"
]

for column in approval_features:

    print("\n======================================")
    print("Approval Rate by", column)
    print("======================================")

    approval_rate = (
        clean_df
        .groupby(column)["Loan_Status"]
        .apply(lambda x: (x == "Y").mean() * 100)
        .sort_values(ascending=False)
    )

    display(
        approval_rate
        .round(2)
        .to_frame("Approval_Rate_%")
    )


# ============================================================
# 62. CROSS TABULATION
# ============================================================

print("\nEducation vs Loan Status:")

education_status = pd.crosstab(
    clean_df["Education"],
    clean_df["Loan_Status"],
    margins=True
)

display(education_status)


print("\nProperty Area vs Loan Status:")

property_status = pd.crosstab(
    clean_df["Property_Area"],
    clean_df["Loan_Status"],
    margins=True
)

display(property_status)


print("\nMarried vs Loan Status:")

married_status = pd.crosstab(
    clean_df["Married"],
    clean_df["Loan_Status"],
    margins=True
)

display(married_status)


# ============================================================
# 63. APPROVAL RATE HEATMAP
# ============================================================

approval_table = pd.crosstab(
    clean_df["Education"],
    clean_df["Property_Area"],
    values=clean_df["Loan_Status"].map({
        "Y": 1,
        "N": 0
    }),
    aggfunc="mean"
) * 100

plt.figure(figsize=(9, 6))

sns.heatmap(
    approval_table,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title(
    "Loan Approval Rate: Education vs Property Area"
)

plt.xlabel("Property Area")
plt.ylabel("Education")

plt.show()


# ============================================================
# 64. LOG TRANSFORMATION FOR SKEWED VARIABLES
# ============================================================

clean_df["LogApplicantIncome"] = np.log1p(
    clean_df["ApplicantIncome"]
)

clean_df["LogTotalIncome"] = np.log1p(
    clean_df["TotalIncome"]
)

clean_df["LogLoanAmount"] = np.log1p(
    clean_df["LoanAmount"]
)


# ============================================================
# 65. BEFORE AND AFTER LOG TRANSFORMATION
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.histplot(
    clean_df["ApplicantIncome"],
    kde=True
)

plt.title("Original Applicant Income")


plt.subplot(1, 2, 2)

sns.histplot(
    clean_df["LogApplicantIncome"],
    kde=True
)

plt.title("Log Applicant Income")

plt.tight_layout()
plt.show()


# ============================================================
# 66. DATASET BEFORE ENCODING
# ============================================================

print("\nClean Dataset:")

display(
    clean_df.head()
)


print("\nClean Dataset Shape:")

print(clean_df.shape)


# ============================================================
# 67. DROP LOAN_ID
# ============================================================

model_df = clean_df.drop(
    columns=["Loan_ID"]
).copy()


# ============================================================
# 68. ENCODE TARGET VARIABLE
# ============================================================

model_df["Loan_Status"] = (
    model_df["Loan_Status"]
    .map({
        "N": 0,
        "Y": 1
    })
)


# ============================================================
# 69. ONE-HOT ENCODING
# ============================================================

encoding_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Property_Area"
]

model_df = pd.get_dummies(
    model_df,
    columns=encoding_columns,
    drop_first=True
)


# ============================================================
# 70. CONVERT BOOLEAN COLUMNS TO INTEGER
# ============================================================

boolean_columns = model_df.select_dtypes(
    include="bool"
).columns

model_df[boolean_columns] = (
    model_df[boolean_columns]
    .astype(int)
)


# ============================================================
# 71. FINAL DATASET CHECK
# ============================================================

print("\nFinal ML Dataset Shape:")
print(model_df.shape)

print("\nFinal Columns:")
print(model_df.columns.tolist())

print("\nMissing Values:")
display(
    model_df.isnull().sum()
    .to_frame("Missing")
)

print("\nData Types:")
display(
    model_df.dtypes.to_frame("Data_Type")
)


# ============================================================
# 72. FINAL DATASET
# ============================================================

print("\nFinal ML Ready Dataset:")

display(
    model_df.head(10)
)


# ============================================================
# 73. SAVE CLEANED DATASET
# ============================================================

clean_df.to_csv(
    "loan_prediction_cleaned.csv",
    index=False
)

print(
    "\nCleaned dataset saved as:"
    " loan_prediction_cleaned.csv"
)


# ============================================================
# 74. SAVE ML READY DATASET
# ============================================================

model_df.to_csv(
    "loan_prediction_model_ready.csv",
    index=False
)

print(
    "ML-ready dataset saved as:"
    " loan_prediction_model_ready.csv"
)


# ============================================================
# 75. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("LOAN PREDICTION PROJECT SUMMARY")
print("=" * 60)

print("Original Dataset Shape :", df.shape)
print("Clean Dataset Shape   :", clean_df.shape)
print("ML Dataset Shape      :", model_df.shape)

print(
    "Duplicate Rows        :",
    df.duplicated().sum()
)

print(
    "Remaining Missing     :",
    clean_df.isnull().sum().sum()
)

print(
    "Approved Loans        :",
    (clean_df["Loan_Status"] == "Y").sum()
)

print(
    "Rejected Loans        :",
    (clean_df["Loan_Status"] == "N").sum()
)

approval_rate = (
    clean_df["Loan_Status"]
    .eq("Y")
    .mean() * 100
)

print(
    "Overall Approval Rate :",
    round(approval_rate, 2),
    "%"
)

print("=" * 60)
print("EDA AND DATA CLEANING COMPLETED!")
print("=" * 60)
```
